# Exploratory Analysis — LFW / Low-light Face Recognition

Optional notebook for visual inspection of raw vs enhanced faces.
Run data preparation first, then execute cells from the project root kernel.

```python
from pathlib import Path
import json
import matplotlib.pyplot as plt
import cv2

ROOT = Path('..').resolve()
splits = json.loads((ROOT / 'data/processed/splits.json').read_text(encoding='utf-8'))
print(splits['counts'], 'classes=', splits['num_classes'])
```

```python
sample = splits['splits']['train'][:6]
fig, axes = plt.subplots(2, 3, figsize=(9, 6))
for ax, item in zip(axes.ravel(), sample):
    img = cv2.imread(str(ROOT / item['path']))
    ax.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
    ax.set_title(('LL' if item['low_light'] else 'NL') + f" | {item['identity'][:12]}")
    ax.axis('off')
plt.tight_layout()
plt.show()
```
